```python
# filtering data using the loc function
import pandas as pd
import string
sportsfile=pd.read_csv("https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/refs/heads/master/data/bios.csv")
print("this the search opperaton for hight criteria:\n")
hightsearch=sportsfile.loc[(sportsfile["height_cm"]>215)&(sportsfile["born_country"]=="MAR"),["name","height_cm","born_country"]].sort_values("height_cm",ascending=False)
print(hightsearch)
print("\n")
print("this the search opperaton for the name criteria:\n")
namesearch=sportsfile.loc[sportsfile["name"].str.contains("Moad",case=False,na=False),["name", "born_country"]]
print(namesearch)
sportsfile['hight category']=sportsfile['height_cm'].apply(lambda x: 'short' if x<165 else('avrage' if x<190 else(pd.NA if pd.isna(x) else 'tall')))
sportsfile.head()
```
